# 3. Define Transfer Learning model
Pytorch has an inbuilt [models package](https://pytorch.org/vision/stable/models.html) that allows loading of popular models with pre-trained weights.

* We want to add an additional classifier stage (to the output of the network). How to setup the [model is detailed here](https://discuss.pytorch.org/t/load-only-a-part-of-the-network-with-pretrained-weights/88397/2).
* This additional classifier may just be a single layer or a cascade of fully connected layers with dropout.
* Note that the number of parameters in the convolutional part of the model will be same no what the input size is set to. However the output feature map size will vary with input image size (small for small image, large for large image). This means the number of parameters in the additional classifier will change depending on input image size.


```python
# Get cpu, gpu or mps device for training.
device = (
    "cuda"
    if torch.cuda.is_available()
    else "mps"
    if torch.backends.mps.is_available()
    else "cpu"
)
print(f"Using {device} device")
```

```python
from torchvision import models
from torchinfo import summary

def set_requires_grad(layer, value):
    """Sets the requires_grad property and returns the layer"""
    #layer.requires_grad = value
    return layer

def remove_classifier_and_fix_weights(layers):
    """
    Build a new nn.Sequential which has no classifier
        and where all of the weights are fixed.
        Modifies layers in place
    layers: nn.Sequential
    """
    
    return nn.Sequential(*[remove_classifier_and_fix_weights(layer) if isinstance(layer, nn.Sequential) 
        else set_requires_grad(layer, False)
        for name, layer in layers.named_children() if name != "classifier"])

class TransferNet(nn.Module):
    def __init__(self, pretrained, n_classes):
        super(TransferNet, self).__init__()

        children = remove_classifier_and_fix_weights(pretrained)

        self.layers = nn.Sequential(
            # take the classification and AvgPooling layers, drop the classification
            children,

            # create new classification head with same size network
            # as original AlexNet. (with different output param size)
            nn.Sequential(
                nn.Flatten(),
                nn.Dropout(p=0.5, inplace=False),
                # nn.Linear(in_features=input_w*input_h, out_features=4096, bias=True),
                # nn.Linear(in_features=18432, out_features=4096, bias=True),
                nn.Linear(in_features=9216, out_features=4096, bias=True),
                nn.ReLU(inplace=True),
                nn.Dropout(p=0.5, inplace=False),
                nn.Linear(in_features=4096, out_features=4096, bias=True),
                nn.ReLU(inplace=True),
                nn.Linear(in_features=4096, out_features=n_classes, bias=True)
        ))
    
    def forward(self, x):
        return self.layers(x)


#print(models.alexnet(weights=models.AlexNet_Weights.DEFAULT))
model = TransferNet(pretrained=models.alexnet(weights=models.AlexNet_Weights.DEFAULT), n_classes=10)
print(model)
print(summary(model, depth=4))

# RuntimeError: mat1 and mat2 shapes cannot be multiplied (3072x6 and 9216x4096)
# RuntimeError: mat1 and mat2 shapes cannot be multiplied (3072x6 and 50176x4096)
# torch.Size([3, 224, 224])
```

# 4. Optimisation & Training Loop

Define a training loop that prints the loss from the training and validation set at least every epoch.
* You may choose to make the validation loss calculations more frequent so you can ensure training is progressing satisfactorily (especially with larger datasets).

* You can use tensorboard to visualise the [loss curves](https://pytorch.org/docs/stable/tensorboard.html))

* For multi-class classification classification problem we will use the  [CrossEntropy loss](https://pytorch.org/docs/stable/generated/torch.nn.CrossEntropyLoss.html). In this pytorch function, the input is the un-normalised logit value.

* You may  have to use [torch.squeeze](https://pytorch.org/docs/stable/generated/torch.squeeze.html) to reduce the dimensionality of the label tensor before passing it to the loss function (this due to how the dataset is configured and the loss will only accept 0D or 1D inputs)

```python
from torch.utils.tensorboard import SummaryWriter
from torch.utils.data import DataLoader
from torch import nn, optim
from datetime import datetime
            
def compile_stats(stats):
    """
    Build loss and accuracy stats from a list of batch results

    stats:
        [ (batch 1 loss, [batch 1/image 1 success, batch 1/image 2 success, ...])],
          (batch 2 loss, [batch 2/image 1 success, batch 2/image 2 success, ...])],
          ...
        ]
    """

    loss, results = zip(*stats)
    avg_loss = torch.as_tensor(loss).mean().item()
    avg_acc = torch.cat(results).type(torch.float).mean().item()
    
    return avg_loss, avg_acc

class Trainer():
    def __init__(self, model, data_train, data_val, device, batch_size=256, lr=0.001, weight_decay=0.0, images_per_checkpoint=1_000):
        """
        images_per_checkpoint:  the number of images to process before emitting
                                and event to the tensorboard writer
        """
        self.device = device
        self.model = model.to(device)
        self.images_per_checkpoint = images_per_checkpoint
        
        self.loss_fn = nn.CrossEntropyLoss()
        self.optimizer = optim.SGD(model.parameters(), lr=lr, weight_decay=weight_decay)
        self.data_train = DataLoader(dataset=data_train, batch_size=batch_size, shuffle=True)
        self.data_val = DataLoader(dataset=data_val, batch_size=batch_size, shuffle=True)
        self.checkpoint = 0
        self.train_stats = []

    def train(self, n_epochs=20):
        """
        Train the entire model
        """

        for e in range(1, n_epochs + 1):
            print(f"Epoch {e}")

            for batch_result in self.train_epoch():
                yield e, batch_result

    def train_epoch(self):
        """
        Train a single epoch
        """

        for data in self.data_train:
            batch_result = self.train_and_log_batch(data)
            if batch_result: yield batch_result
            
    def train_and_log_batch(self, batch):
        """
        Train a single batch and return a training set result
        if required (determined by images_per_checkpoint)
        """

        loss, results = self.train_batch(batch)
        self.train_stats.append((loss, results))
        self.checkpoint += len(results)
        
        return self.compile_and_reset_train_stats()\
            if self.checkpoint >= self.images_per_checkpoint\
            else None
            
    def compile_and_reset_train_stats(self):
        """
        Compile a training set result and reset totals to 0
        """

        stats = compile_stats(self.train_stats)
        ch = self.checkpoint

        self.train_stats = []
        self.checkpoint = 0
        
        return ch, stats
            
    def train_batch(self, batch):
        """
        Train on a batch, updating gradients and return results
        """

        result = self.execute_batch(batch)
        self.optimizer.step()
        self.optimizer.zero_grad()

        return result
            
    def execute_batch(self, batch):
        """
        Execute a single batch and return results
        """

        inputs, labels = batch
        inputs, labels = inputs.to(self.device), labels[:,0].to(self.device)
        outputs = self.model(inputs)

        loss = self.loss_fn(outputs, labels)
        results = outputs.argmax(1) == labels

        return loss, results

    def validate(self):
        """
        Execute the entire validation data set and return loss and acc stats
        """

        self.model.train(False)
        try:
            stats = map(lambda x: self.execute_batch(x), self.data_val)
            return compile_stats(stats)
        
        finally:
            self.model.train(True)

class LoggedTrainer():
    def __init__(self, name, trainer):
        """
        images_per_checkpoint:  the number of images to process before emitting
                                and event to the tensorboard writer
        """
        self.writer = SummaryWriter(f'tensorboard/{round(datetime.now().timestamp())}')
        self.name = name
        self.trainer = trainer
        self.total_images = 0

    def __enter__(self):
        return self
    
    def __exit__(self, exception_type, exception_value, exception_traceback):
        self.writer.close()

    def train(self, n_epochs=20):
        """
        Train the entire model and emit results to tensorboard
        """

        for epoch, event in self.trainer.train(n_epochs=n_epochs):
            test_count, (train_loss, train_acc) = event
            val_loss, val_acc = self.trainer.validate()
            self.write(epoch, test_count, train_loss, train_acc, val_loss, val_acc)

    def write(self, epoch, test_count, train_loss, train_acc, val_loss, val_acc):
        """
        Write test/validation stats to tensorboard
        """

        self.total_images += test_count
        
        print(f"Training Error, epoch {epoch}: \n Accuracy: {(train_acc):>0.1f}%, Avg loss: {train_loss:>8f}")
        print(f"Validation Error, epoch {epoch}: \n Accuracy: {(val_acc):>0.1f}%, Avg loss: {val_loss:>8f}\n")

        # Log the running loss averaged per batch
        self.writer.add_scalars('Loss',
                        { f'Training Loss {self.name}' : train_loss, f'Validation Loss {self.name}' : val_loss },
                        self.total_images)
        self.writer.add_scalars('Accuracy',
                        {f'Training Acc {self.name}' : train_acc, f'Validation Acc {self.name}' : val_acc },
                        self.total_images)

        self.writer.flush()

with LoggedTrainer(
    name="x",
    trainer=Trainer(
        model=TransferNet(
            n_classes=8,
            pretrained=models.alexnet(weights=models.AlexNet_Weights.DEFAULT)),
        data_train=train_dataset, 
        data_val=val_dataset,
        device=device,
        batch_size=256,
        images_per_checkpoint=1_000)) as trainer:

    trainer.train(5)

        
        
```
